**Business:** Wanjiku's Fresh-Produce Stall  
**Analysis Period:** Two Weeks (14 Days)  
**Currency:** Kenyan Shillings (KES)

## Introduction

Wanjiku operates a fresh-produce stall at Gikomba Market.
She has recorded her daily sales for two weeks and maintains
a basic stock record.

The purpose of this analysis is to evaluate her business
performance, identify sales trends, classify daily performance,
compare weekday and weekend sales, assess inventory levels,
and provide practical recommendations to improve profitability
and stock management.

This analysis uses core Python data structures, functions,
loops, conditional statements and formatted output.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg

stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

### Approach

We calculate the total sales by adding all daily sales values.
The average is obtained by dividing total sales by 14 days.
The maximum and minimum functions identify the highest and
lowest sales values.

We then use the index() method to locate the positions of
these values in the daily_sales list. Integer division by 7
helps identify the corresponding week, while the days list
provides the day names.

```python
# Calculate overall sales statistics

total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)

highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

# Find the positions of the highest and lowest sales
best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

# Identify the corresponding days and weeks
best_day = days[best_index]
worst_day = days[worst_index]

best_week = (best_index // 7) + 1
worst_week = (worst_index // 7) + 1

# Display results
print("OVERALL SALES PERFORMANCE")
print(f"Total Sales:       KES {total_sales:,.2f}")
print(f"Average Daily:     KES {average_sales:,.2f}")
print(f"Highest Sales:     KES {highest_sales:,.2f}")
print(f"Lowest Sales:      KES {lowest_sales:,.2f}")
print(f"Best Sales Day:    {best_day}, Week {best_week}")
print(f"Worst Sales Day:   {worst_day}, Week {worst_week}")
```

## Task 2: Weekly Sales and Growth

### Approach

We define a reusable function called week_totals(sales).
The function uses list slicing to separate the first seven
days from the next seven days.

The sum() function calculates the total sales for each week.
We then calculate the week-on-week growth percentage by
comparing Week 2 sales against Week 1 sales.

A positive percentage indicates growth, while a negative
percentage indicates a decline.

```python
def week_totals(sales):
    """
    Calculate and return the total sales for Week 1 and Week 2.

    Parameters:
        sales (list): Daily sales values for 14 days.

    Returns:
        tuple: Total sales for Week 1 and Week 2.
    """
    week1 = sum(sales[:7])
    week2 = sum(sales[7:14])

    return week1, week2


# Calculate weekly totals
week1_total, week2_total = week_totals(daily_sales)

# Calculate week-on-week growth
growth_percentage = ((week2_total - week1_total) / week1_total) * 100

print("WEEKLY SALES ANALYSIS")
print(f"Week 1 Sales:      KES {week1_total:,.2f}")
print(f"Week 2 Sales:      KES {week2_total:,.2f}")
print(f"Sales Growth:      {growth_percentage:.2f}%")
```

## Task 3: Daily Sales Classification

### Approach

We classify each day's sales into three categories using
the average daily sales as our benchmark.

- Strong: Sales are more than 20% above average.
- Weak: Sales are more than 20% below average.
- Normal: Sales fall within the remaining range.

We create a function called classify_day(amount, average)
using conditional statements.

We then apply this function to every daily sales value
and use a dictionary to count the number of days in
each category.

```python
def classify_day(amount, average):
    """
    Classify daily sales as Strong, Normal or Weak.

    Parameters:
        amount (float): Sales recorded for a particular day.
        average (float): Average daily sales.

    Returns:
        str: Sales performance category.
    """
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


# Classify sales for every day
daily_categories = []

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    daily_categories.append(category)


# Count the number of days in each category
category_counts = {
    "Strong": 0,
    "Normal": 0,
    "Weak": 0
}

for category in daily_categories:
    category_counts[category] += 1


# Display daily classifications
print("DAILY SALES CLASSIFICATION")

for i in range(len(daily_sales)):
    print(f"{days[i]} (Day {i + 1}): "
          f"KES {daily_sales[i]:,.2f} - {daily_categories[i]}")


# Display category totals
print("\nCATEGORY SUMMARY")

for category, count in category_counts.items():
    print(f"{category}: {count} days")
```

## Task 4: Weekday versus Weekend Performance

### Approach

We separate sales into two groups:

- Weekdays: Monday to Friday.
- Weekends: Saturday and Sunday.

Using list slicing, we extract the first five days and
the last two days of each week.

We calculate the average sales for each group and
compare them using a percentage difference.

The percentage difference is calculated relative to
the weekday average.

```python
# Separate weekday and weekend sales for both weeks

week1_weekdays = daily_sales[:5]
week1_weekends = daily_sales[5:7]

week2_weekdays = daily_sales[7:12]
week2_weekends = daily_sales[12:14]

# Combine both weeks
weekday_sales = week1_weekdays + week2_weekdays
weekend_sales = week1_weekends + week2_weekends

# Calculate average sales
weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

# Calculate percentage difference
weekend_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

print("WEEKDAY VERSUS WEEKEND PERFORMANCE")
print(f"Weekday Average:  KES {weekday_average:,.2f}")
print(f"Weekend Average:  KES {weekend_average:,.2f}")
print(f"Difference:       {weekend_difference:.2f}%")

if weekend_average > weekday_average:
    print("Weekend sales perform better than weekday sales.")
elif weekend_average < weekday_average:
    print("Weekday sales perform better than weekend sales.")
else:
    print("Weekday and weekend sales are equal.")
```

## Task 5: Stock Reordering and Gross Margins

### Approach

We examine the stock dictionary to identify products whose
current quantity is at or below their reorder levels.

We create a reusable function called products_to_reorder()
that loops through the stock dictionary using .items().

We also calculate the gross margin percentage for every
product using the formula:

Gross Margin = ((Selling Price - Cost Price) / Selling Price) × 100

This helps Wanjiku understand the profitability of each
product and identify which products require restocking.

```python
def products_to_reorder(stock_data):
    """
    Return a list of products whose stock quantity is
    at or below their specified reorder level.

    Parameters:
        stock_data (dict): Dictionary containing product
                           quantities and reorder levels.

    Returns:
        list: Names of products requiring restocking.
    """
    reorder_list = []

    for product, details in stock_data.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


# Identify products requiring restocking
reorder_products = products_to_reorder(stock)


# Calculate gross margins for all products
gross_margins = {}

for product, details in stock.items():
    cost = details["cost"]
    price = details["price"]

    margin = ((price - cost) / price) * 100

    gross_margins[product] = margin


# Display stock analysis
print("STOCK AND PROFITABILITY ANALYSIS")

print(f"{'Product':<15}{'Quantity':>10}{'Reorder':>10}"
      f"{'Margin':>12}{'Status':>15}")


for product, details in stock.items():

    status = "REORDER" if product in reorder_products else "Sufficient"

    print(f"{product:<15}{details['qty']:>10}"
          f"{details['reorder']:>10}"
          f"{gross_margins[product]:>11.2f}%"
          f"{status:>15}")


print("\nPRODUCTS REQUIRING RESTOCKING")

for product in reorder_products:
    print(product)
```
