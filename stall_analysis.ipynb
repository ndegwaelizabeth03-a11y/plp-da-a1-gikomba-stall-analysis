# Gikomba Market Stall Performance Analysis
## Dataset

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]   # 14 days, KES
days = ["Mon", "Tue", "Wed", "Thu", "Fri", 'Sat', 'Sun'] * 2
# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Sales summary
sum/max/min, `list.index()`; week = position // 7 + 1.

```python
total = sum(daily_sales)
average = total / len(daily_sales)
high, low = max(daily_sales), min(daily_sales)
hi_pos, lo_pos = daily_sales.index(high), daily_sales.index(low)
```

## Task 2: Weekly totals and growth
Slice two weeks, return a tuple, get growth %.

```python
def week_totals(sales):
    '''Return (week 1 total, week 2 total).'''
    return sum(sales[:7]), sum(sales[7:14])

wk1, wk2 = week_totals(daily_sales)
growth = (wk2 - wk1) / wk1 * 100
```

## Task 3: Classify days
20% above/below average; dict counts classes.

```python
def classify_day(amount, average):
    '''Return Strong (>20% above average), Weak (>20% below) or Normal.'''
    if amount > average * 1.2:
        return 'Strong'
    if amount < average * 0.8:
        return 'Weak'
    return 'Normal'

counts = {}
for amount in daily_sales:
    label = classify_day(amount, average)
    counts[label] = counts.get(label, 0) + 1
```

## Task 4: Weekday vs weekend
Average each group; gap as % of weekday.

```python
weekday = [s for d, s in zip(days, daily_sales) if d not in ('Sat', 'Sun')]
weekend = [s for d, s in zip(days, daily_sales) if d in ('Sat', 'Sun')]
wd_avg, we_avg = sum(weekday) / len(weekday), sum(weekend) / len(weekend)
diff_pct = (we_avg - wd_avg) / wd_avg * 100
```

## Task 5: Reorder list and margins
Loop with `.items()`; margin = (price-cost)/price.

```python
def products_to_reorder(stock_data):
    '''Return products at or below the reorder level.'''
    return [p for p, d in stock_data.items() if d['qty'] <= d['reorder']]

reorder = products_to_reorder(stock)
margins = {p: (d['price'] - d['cost']) / d['price'] * 100 for p, d in stock.items()}
```

## Task 6: Stall Performance Report
f-strings, KES with thousands separators.

```python
print('STALL PERFORMANCE REPORT')
print(f'Total sales: KES {total:,}')
print(f'Average/day: KES {average:,.2f}')
print(f'Best day: KES {high:,} ({days[hi_pos]}, wk {hi_pos // 7 + 1})')
print(f'Worst day: KES {low:,} ({days[lo_pos]}, wk {lo_pos // 7 + 1})')
print(f'Week 1/2: KES {wk1:,} / KES {wk2:,} ({growth:.2f}%)')
print(f'Strong/Normal/Weak: {counts["Strong"]}/{counts["Normal"]}/{counts["Weak"]}')
print(f'Weekday avg: KES {wd_avg:,.2f}')
print(f'Weekend avg: KES {we_avg:,.2f} (+{diff_pct:.2f}%)')
print('Reorder:', ', '.join(reorder))
for p, d in stock.items():
    print(f'{p:<9}{d["qty"]:>3} kg {margins[p]:>6.2f}%')
```
